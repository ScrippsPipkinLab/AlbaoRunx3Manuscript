## Convert EnsemblIDs to Gene Symbols+

```r
# Load EnsemblID conversion
ensembl2symbol <- read.table("source_data/ensembl_102.txt", header = TRUE, sep = "\t", stringsAsFactors = FALSE)
colnames(ensembl2symbol) <- c("GeneID", "GeneSymbol")

# Check df
head(ensembl2symbol)
```

```r
# Function to convert EnsemblIDs to GeneSymbols
convert_ids <- function(ids, conversion_df) {
    sapply(ids, function(id) {
        id_list <- unlist(strsplit(id, ";"))
        symbol_list <- sapply(id_list, function(single_id) {
            symbol <- conversion_df$GeneSymbol[conversion_df$GeneID == single_id]
            if (length(symbol) == 0 || is.na(symbol[1]) || symbol[1] == "") {
                return(single_id)
            } else {
                return(symbol)
            }
        })
        paste(symbol_list, collapse = ";")
    })
}
```

```r
# Keep only items in ensemble2symbol that are in peak_anno_df
uniqueids <- unique(unlist(strsplit(peaks$flank_geneIds, ";")))
uniqueids <- unique(c(uniqueids, peaks$flank_geneId))

ensembl2symbol <- ensembl2symbol[ensembl2symbol$GeneID %in% uniqueids, ]
```

```r
# Convert geneId and flank_geneIds columns
peaks$geneSymbol <- convert_ids(peaks$geneId, ensembl2symbol)
peaks$flank_geneSymbols <- convert_ids(peaks$flank_geneIds, ensembl2symbol)
print("Done Annotation")
```

## Add Motifs

```r
# Load motif extraction results
motifs <- read.table("03_motif/motifAnnotated.simple.tsv", header = FALSE, sep = "\t", stringsAsFactors = FALSE)
colnames(motifs) <- c("PositionID", "Motif")

head(motifs)
```

```r
# Collapse motifs by PositionID
motifs <- motifs %>%
    group_by(PositionID) %>%
    summarise(Motif = paste(unique(Motif), collapse = ";"))

library(stringr)

extract_first_field <- function(x) {
    items <- str_split(x, ";")[[1]]              # split into individual motif entries
    first_fields <- sapply(items, function(i) str_split(i, ",")[[1]][1])  # take first field of each
    paste(first_fields, collapse = ",")
}

motifs <- motifs %>%
  mutate(MotifNames = sapply(Motif, extract_first_field))

head(motifs)
```

```r
# Left join motifs to peaks, use V4 in peaks and PositionID in motifs
peaks <- left_join(peaks, motifs, by = c("V4" = "PositionID"))
head(peaks)

# Replace V4 column name with PeakID
colnames(peaks)[colnames(peaks) == "V4"] <- "PeakID"
# Replace V5 with Cluster
colnames(peaks)[colnames(peaks) == "V5"] <- "Cluster"

head(peaks)
```

```r
# Save CSV and RDS
write.csv(peaks, "04_peakAnnotation/peaks.annotated.csv", row.names = FALSE)
saveRDS(peaks, "04_peakAnnotation/peaks.annotated.rds")
```
