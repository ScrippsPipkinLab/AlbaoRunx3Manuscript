```bash
# Define input bigWig files
basePath="/home/dalbao/AlbaoRunx3Manuscript/cutnrun/source_data/260713_reBAM2_noDeDup_CPM/04_reporting/igv/"

declare -A inputBigwig

for target in Runx3 Runx1 IgG; do
    for group in shCd19 shRunx3 memory early late terminal; do
        name="${group}_${target}"
        inputBigwig[$name]="${basePath}${group}_${target}_R1.bigWig"
        if [ ! -f "${inputBigwig[$name]}" ]; then
            echo "Error: Input file ${inputBigwig[$name]} does not exist."
            exit 1
        fi
        echo ${inputBigwig[$name]}
    done
done
```

## 04.02 Plot Raw
Plot RUNX1 and RUNX3 binding on shCd19 and shRunx3 (tautological, but a good control)

```bash
mkdir -p 04_peakheatmap/raw_maps

for peakset in Runx3 Runx1; do
    for signal in Runx3 Runx1; do
        for day in day5 day8; do
            deeptools computeMatrix reference-point \
                -S \
                ${inputBigwig["shCd19_${signal}"]} \
                ${inputBigwig["shRunx3_${signal}"]} \
                ${inputBigwig["shCd19_IgG"]} \
                ${inputBigwig["shRunx3_IgG"]} \
                -R \
                "03_peakconsolidation/${day}_${peakset}.bound.bed" \
                "03_peakconsolidation/${day}_${peakset}.constant.bed" \
                "03_peakconsolidation/${day}_${peakset}.unknown.bed" \
                --referencePoint center \
                -b 1500 -a 1500 \
                --numberOfProcessors 36 \
                --sortUsing sum \
                -out "04_peakheatmap/raw_maps/matrix/perturb_${signal}Signal_${peakset}PeakSet_${day}_raw.gz"
        done
    done
done

for peakset in Runx3 Runx1; do
    for signal in Runx3 Runx1; do
        for day in day5 day8; do
            deeptools computeMatrix reference-point \
                -S \
                ${inputBigwig["memory_${signal}"]} \
                ${inputBigwig["early_${signal}"]} \
                ${inputBigwig["late_${signal}"]} \
                ${inputBigwig["terminal_${signal}"]} \
                ${inputBigwig["memory_IgG"]} \
                ${inputBigwig["early_IgG"]} \
                ${inputBigwig["late_IgG"]} \
                ${inputBigwig["terminal_IgG"]} \
                -R \
                "03_peakconsolidation/${day}_${peakset}.bound.bed" \
                "03_peakconsolidation/${day}_${peakset}.constant.bed" \
                "03_peakconsolidation/${day}_${peakset}.unknown.bed" \
                --referencePoint center \
                -b 1500 -a 1500 \
                --numberOfProcessors 36 \
                --sortUsing sum \
                -out "04_peakheatmap/raw_maps/matrix/natural_${signal}Signal_${peakset}PeakSet_${day}_raw.gz"
        done
    done
done
```

```bash
mkdir -p 04_peakheatmap/raw_maps/plots

for peakset in Runx3 Runx1; do
    for signal in Runx3 Runx1; do
        for day in day5 day8; do
            for experiment in perturb natural; do
                deeptools plotHeatmap \
                    -m "04_peakheatmap/raw_maps/matrix/${experiment}_${signal}Signal_${peakset}PeakSet_${day}_raw.gz" \
                    -out "04_peakheatmap/raw_maps/plots/${experiment}_${signal}Signal_${peakset}PeakSet_${day}_raw.pdf" \
                    --sortUsing sum \
                    --colorMap "RdYlBu" \
                    --regionsLabel "${peakset}-Bound" "${peakset}-Constant" "${peakset}-Unknown"
            done
        done
    done
done
```

## 04.03 Background Correction

The day 8 Runx3 heatmap shows an elevated background in the terminal timepoint relative to memory/early/late, even outside called peaks. CPM normalization only corrects for sequencing depth, not this kind of sample-to-sample noise difference. Use `bigwigCompare` to subtract each timepoint's own matched IgG track from its Runx3 track (both `subtract` and `log2` ratio), then replot on the IgG-corrected signal.

```bash
mkdir -p 04_peakheatmap/background_corrected

declare -A correctedSubtract
declare -A correctedLog2

for group in memory early late terminal shCd19 shRunx3; do
    for signal in Runx3 Runx1; do
        name="${group}_${signal}"
        subtractOut="04_peakheatmap/background_corrected/${name}_subtract.bigWig"
        log2Out="04_peakheatmap/background_corrected/${name}_log2.bigWig"

        # deeptools bigwigCompare \
        #     -b1 "${inputBigwig[$name]}" \
        #     -b2 "${inputBigwig["${group}_IgG"]}" \
        #     --operation subtract \
        #     --numberOfProcessors 36 \
        #     -o "$subtractOut"

        # deeptools bigwigCompare \
        #     -b1 "${inputBigwig[$name]}" \
        #     -b2 "${inputBigwig["${group}_IgG"]}" \
        #     --operation log2 \
        #     --numberOfProcessors 36 \
        #     -o "$log2Out"

        correctedSubtract[$name]="$subtractOut"
        correctedLog2[$name]="$log2Out"
    done
done
```

## 04.04 Plot Corrected

```bash
mkdir -p 04_peakheatmap/corrected_maps 04_peakheatmap/corrected_maps/subtract 04_peakheatmap/corrected_maps/log2

for operation in subtract log2; do
    if [ "$operation" = "subtract" ]; then
        declare -n corrected=correctedSubtract
    else
        declare -n corrected=correctedLog2
    fi
    for signal in Runx3 Runx1; do
        for peakset in Runx3 Runx1; do
            for day in day5 day8; do
                deeptools computeMatrix reference-point \
                    -S \
                    ${corrected["memory_${signal}"]} \
                    ${corrected["early_${signal}"]} \
                    ${corrected["late_${signal}"]} \
                    ${corrected["terminal_${signal}"]} \
                    -R \
                    "03_peakconsolidation/${day}_${peakset}.bound.bed" \
                    "03_peakconsolidation/${day}_${peakset}.constant.bed" \
                    "03_peakconsolidation/${day}_${peakset}.unknown.bed" \
                    --referencePoint center \
                    -b 1500 -a 1500 \
                    --numberOfProcessors 36 \
                    --sortUsing sum \
                    -out "04_peakheatmap/corrected_maps/matrix/natural_${signal}Signal_${peakset}PeakSet_${day}_${operation}.gz"

                deeptools computeMatrix reference-point \
                    -S \
                    ${corrected["shCd19_${signal}"]} \
                    ${corrected["shRunx3_${signal}"]} \
                    -R \
                    "03_peakconsolidation/${day}_${peakset}.bound.bed" \
                    "03_peakconsolidation/${day}_${peakset}.constant.bed" \
                    "03_peakconsolidation/${day}_${peakset}.unknown.bed" \
                    --referencePoint center \
                    -b 1500 -a 1500 \
                    --numberOfProcessors 36 \
                    --sortUsing sum \
                    -out "04_peakheatmap/corrected_maps/matrix/perturb_${signal}Signal_${peakset}PeakSet_${day}_${operation}.gz"
            done
        done
    done
done
```

```bash
for operation in subtract log2; do
    for target in Runx3 Runx1; do
        for peakset in Runx3 Runx1; do
            for day in day5 day8; do
                for experiment in perturb natural; do
                    # These are SAMPLE (track) names → --samplesLabel
                    if [ "$experiment" = "perturb" ]; then
                        samples=("shCd19_${target}" "shRunx3_${target}")
                    else
                        samples=("memory_${target}" "early_${target}" "late_${target}" "terminal_${target}")
                    fi

                    # Region groups come from the -R BEDs: bound / constant / unknown
                    regions=("${peakset}-Bound" "${peakset}-Constant" "${peakset}-Unknown")

                    deeptools plotHeatmap \
                        -m   "04_peakheatmap/corrected_maps/matrix/${experiment}_${target}Signal_${peakset}PeakSet_${day}_${operation}.gz" \
                        -out "04_peakheatmap/corrected_maps/${operation}/${experiment}_${target}Signal_${peakset}PeakSet_${day}_${operation}.pdf" \
                        --sortUsing max \
                        --colorMap "RdBu_r" \
                        --samplesLabel "${samples[@]}" \
                        --regionsLabel "${regions[@]}"
                done
            done
        done
    done
done
```

## 04.05 Clustering

```bash
mkdir -p 04_peakheatmap/clustering/matrix 04_peakheatmap/clustering/subtract 04_peakheatmap/clustering/log2

for operation in subtract log2; do
    if [ "$operation" = "subtract" ]; then
        declare -n corrected=correctedSubtract
    else
        declare -n corrected=correctedLog2
    fi
    for day in day5 day8; do
        deeptools computeMatrix reference-point \
            -S \
            ${corrected["memory_Runx3"]} \
            ${corrected["early_Runx3"]} \
            ${corrected["late_Runx3"]} \
            ${corrected["terminal_Runx3"]} \
            ${corrected["memory_Runx1"]} \
            ${corrected["early_Runx1"]} \
            ${corrected["late_Runx1"]} \
            ${corrected["terminal_Runx1"]} \
            -R \
            "03_peakconsolidation/${day}_Runx1_Runx3.union.bed" \
            --referencePoint center \
            -b 1500 -a 1500 \
            --numberOfProcessors 36 \
            --sortUsing sum \
            -out "04_peakheatmap/clustering/matrix/natural_clustering_${day}_${operation}.gz"

        deeptools computeMatrix reference-point \
            -S \
            ${corrected["shCd19_Runx3"]} \
            ${corrected["shRunx3_Runx3"]} \
            ${corrected["shCd19_Runx1"]} \
            ${corrected["shRunx3_Runx1"]} \
            -R \
            "03_peakconsolidation/${day}_Runx1_Runx3.union.bed" \
            --referencePoint center \
            -b 1500 -a 1500 \
            --numberOfProcessors 36 \
            --sortUsing sum \
            -out "04_peakheatmap/clustering/matrix/perturb_clustering_${day}_${operation}.gz"
    done
done
```

```bash
for operation in subtract log2; do
    for day in day5 day8; do
        for experiment in perturb natural; do
            # These are SAMPLE (track) names → --samplesLabel
            if [ "$experiment" = "perturb" ]; then
                samples=("shCd19_Runx3" "shRunx3_Runx3" "shCd19_Runx1" "shRunx3_Runx1")
            else
                samples=("memory_Runx3" "early_Runx3" "late_Runx3" "terminal_Runx3" "memory_Runx1" "early_Runx1" "late_Runx1" "terminal_Runx1")
            fi

            deeptools plotHeatmap \
                -m   "04_peakheatmap/clustering/matrix/${experiment}_clustering_${day}_${operation}.gz" \
                -out "04_peakheatmap/clustering/${operation}/${experiment}_clustering_${day}_${operation}.pdf" \
                --sortUsing max \
                --colorMap "RdBu_r" \
                --samplesLabel "${samples[@]}" \
                --outFileSortedRegions "04_peakheatmap/clustering/${operation}/${experiment}_clustering_${day}_${operation}_clusters.bed" \
                --kmeans 6

        done
    done
done
```
