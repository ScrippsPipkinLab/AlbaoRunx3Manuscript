## 01.2 Initial Heatmaps
Do heatmaps of peaks to analyze.

```bash
# First make a associative array to store bigWig and peak locations for each sample
declare -A bigWigFiles
declare -A rawPeaks

for group in shCd19 shRunx3 memory early late terminal; do
    for target in Runx3 Runx1; do
        fn=${group}_${target}_log2.bigWig
        fpath="source_data/bg_corrected_bigWigs/"
        bigWigFiles[${group}_${target}]="${fpath}${fn}"

        fn=${group}_${target}.CPM.stringent.maxsignal.bed
        fpath="source_data/cpm_stringent_max_peaks/"
        rawPeaks[${group}_${target}]="${fpath}${fn}"
    done
done

echo Sample: ${bigWigFiles["shCd19_Runx3"]}
echo Sample: ${rawPeaks["shCd19_Runx3"]}
```

Check perturbation experiment:

```bash
mkdir -p 01_peakEDA/02_initialHeatmaps

deeptools computeMatrix reference-point \
    -S \
    ${bigWigFiles["shCd19_Runx3"]} \
    ${bigWigFiles["shRunx3_Runx3"]} \
    ${bigWigFiles["shCd19_Runx1"]} \
    ${bigWigFiles["shRunx3_Runx1"]} \
    -R \
    ${rawPeaks["shCd19_Runx3"]} \
    ${rawPeaks["shRunx3_Runx3"]} \
    --referencePoint center \
    -b 1500 -a 1500 \
    --numberOfProcessors 36 \
    --sortUsing mean \
    -out 01_peakEDA/02_initialHeatmaps/perturbSamples.Runx3target-day5peaks.gz

deeptools computeMatrix reference-point \
    -S \
    ${bigWigFiles["shCd19_Runx3"]} \
    ${bigWigFiles["shRunx3_Runx3"]} \
    ${bigWigFiles["shCd19_Runx1"]} \
    ${bigWigFiles["shRunx3_Runx1"]} \
    -R \
    ${rawPeaks["shCd19_Runx1"]} \
    ${rawPeaks["shRunx3_Runx1"]} \
    --referencePoint center \
    -b 1500 -a 1500 \
    --numberOfProcessors 36 \
    --sortUsing mean \
    -out 01_peakEDA/02_initialHeatmaps/perturbSamples.Runx1target-day5peaks.gz
```

# 01.4 Peak Cleanup 

Isolate day 5 good peaks

```bash
mkdir -p 01_peakEDA/04_peakCleanup

# cluster_1 and cluster_2 of Runx3 and Runx1 k4 beds are good.
# Do Runx3 first
cat 01_peakEDA/02_initialHeatmaps/day5peaks_clustering/perturbSamples.Runx3target-day5peaks.k4.sorted.bed \
    | grep "cluster_1\|cluster_2\|cluster_3" \
    > 01_peakEDA/04_peakCleanup/Runx3.good.maxsignal.bed
# Replace cluster_1 with Runx3C1 and cluster_2 with Runx3C2
sed -i 's/cluster_1/Runx3C1/g' 01_peakEDA/04_peakCleanup/Runx3.good.maxsignal.bed
sed -i 's/cluster_2/Runx3C2/g' 01_peakEDA/04_peakCleanup/Runx3.good.maxsignal.bed
sed -i 's/cluster_3/Runx3C3/g' 01_peakEDA/04_peakCleanup/Runx3.good.maxsignal.bed

# Do Runx1 next
cat 01_peakEDA/02_initialHeatmaps/day5peaks_clustering/perturbSamples.Runx1target-day5peaks.k4.sorted.bed \
    | grep "cluster_1\|cluster_2" \
    > 01_peakEDA/04_peakCleanup/Runx1.good.maxsignal.bed
# Replace cluster_1 with Runx1C1 and cluster_2 with Runx1C2
sed -i 's/cluster_1/Runx1C1/g' 01_peakEDA/04_peakCleanup/Runx1.good.maxsignal.bed
sed -i 's/cluster_2/Runx1C2/g' 01_peakEDA/04_peakCleanup/Runx1.good.maxsignal.bed
```

Isolate day 8 peaks

```bash
# In the folder 01_peakEDA/03_peakCleanup are four folders:
# memory early late terminal
# Each folder has a file named Runx3-group.k3.sorted.bed
# Wherein group is the name of the folder (Runx3-early.k3.sorted.bed, Runx3-late.k3.sorted.bed...)
# Write a script that reads each file
# Takes only lines with cluster_1
# And outputs into the new file 04_peakCleanup/group.good.maxsignal.bed

for group in memory early late terminal; do
    input_file="01_peakEDA/03_peakCleanup/${group}/Runx3-${group}.k3.sorted.bed"
    output_file="01_peakEDA/04_peakCleanup/${group}.good.maxsignal.bed"
    
    # Extract lines with cluster_1 and write to the output file
    grep "cluster_1" "$input_file" > "$output_file"
    
    # Optionally, you can replace cluster_1 with a more descriptive name if needed
    sed -i "s/cluster_1/${group}C1/g" "$output_file"
done
```

```bash
# Combine the bed files into combined with the header:
# #chrom	start	end	name	score	strand	thickStart	thickEnd	itemRGB	blockCount	blockSizes	blockStart	deepTools_group
echo '#chrom start end name score strand thickStart thickEnd itemRGB blockCount blockSizes blockStart deepTools_group' > 01_peakEDA/04_peakCleanup/all_good.maxsignal.bed

cat 01_peakEDA/04_peakCleanup/Runx3.good.maxsignal.bed \
    01_peakEDA/04_peakCleanup/Runx1.good.maxsignal.bed \
    01_peakEDA/04_peakCleanup/memory.good.maxsignal.bed \
    01_peakEDA/04_peakCleanup/early.good.maxsignal.bed \
    01_peakEDA/04_peakCleanup/late.good.maxsignal.bed \
    01_peakEDA/04_peakCleanup/terminal.good.maxsignal.bed \
    >> 01_peakEDA/04_peakCleanup/all_good.maxsignal.bed
```

```bash
# Combine all bed files in source/cpm_stringent_max_peaks into one bed file
# save into 04_peakCleanup/broadPeaks.bed
cat source_data/cpm_stringent_peaks/*.bed > 01_peakEDA/04_peakCleanup/broadPeaks.bed

# Use broadPeaks.bed to recover the true peak locations for the good peaks in all_good.maxsignal.bed
# To do this, match column 4 of all_good.maxsignal.bed with column 6 of broadPeaks.bed
# As an additional check. Remember to remove the header from all_good.maxsignal.bed before matching
tail -n +2 01_peakEDA/04_peakCleanup/all_good.maxsignal.bed \
    | awk '{print $4}' \
    | sort \
    | uniq \
    > 01_peakEDA/04_peakCleanup/all_good.maxsignal.names.txt

# Recover true peak locations: keep broadPeaks.bed rows whose column 6 matches a good peak name
awk 'NR==FNR {names[$1]=1; next} $6 in names' \
    01_peakEDA/04_peakCleanup/all_good.maxsignal.names.txt \
    01_peakEDA/04_peakCleanup/broadPeaks.bed \
    > 01_peakEDA/04_peakCleanup/all_good.truePeaks.bed

# QC check: names in all_good.maxsignal.bed that had NO match in broadPeaks.bed
awk 'NR==FNR {names[$1]=1; next} {seen[$6]=1} END {for (n in names) if (!(n in seen)) print n}' \
    01_peakEDA/04_peakCleanup/all_good.maxsignal.names.txt \
    01_peakEDA/04_peakCleanup/broadPeaks.bed \
    > 01_peakEDA/04_peakCleanup/all_good.unmatched_names.txt

# Sanity check counts
echo "Names to match:      $(wc -l < 01_peakEDA/04_peakCleanup/all_good.maxsignal.names.txt)"
echo "Matched broad peaks: $(wc -l < 01_peakEDA/04_peakCleanup/all_good.truePeaks.bed)"
echo "Unmatched names:     $(wc -l < 01_peakEDA/04_peakCleanup/all_good.unmatched_names.txt)"
```
